Level 1 — Just count paths (no grid values at all)
You're at (0,0) on an m×n grid. You can only move right or down. How many distinct paths are there to reach (m-1,n-1)?

```python
def count_paths(m,n,i,j):

    if i == m-1 and j == n-1 :
        return 1

    total = 0

    if i<m-1 :
        total = total + count_paths(m,n,i+1,j)
    if j<n-1 :
        total = total + count_paths(m,n,i,j+1)
    
    return total 

print(count_paths(3,3,0,0))
```

Now applying memoization

```python
def count_paths_dp(m,n,i,j,dp) :
    if i == m-1 and j == n-1 :
        return 1
    if dp[i][j] != -1 :
        return dp[i][j] 
    
    total = 0

    if i<m-1 :
        total = total +count_paths_dp(m,n,i+1,j,dp)
    
    if j<n-1 :
        total = total + count_paths_dp(m,n,i,j+1,dp)
    
    dp[i][j] = total 

    return total

m,n = 3,3 

dp = [[-1]*n for _ in range(m)]

print(count_paths_dp(m,n,0,0,dp))
```

Level 2 — Sum along the path, but just report one total

Same grid movement. Each cell has a value g[i][j]. As you move, you add up the values of cells you land on. What is the sum of values for one arbitrary root-to-leaf path (say, print the path and its sum)?

```python
def count_paths_lvl2(m, n, i, j, path, path_sum):
    if i == m-1 and j == n-1:
        print("sum:", path_sum, " path:", path)
        return

    if i < m-1:
        count_paths_lvl2(m, n, i+1, j, path + [g[i][j]], path_sum + g[i][j])

    if j < n-1:
        count_paths_lvl2(m, n, i, j+1, path + [g[i][j]], path_sum + g[i][j])

g = [[1,2],[3,4]]
m, n = 2, 2
count_paths_lvl2(m, n, 0, 0, [g[0][0]], g[0][0])
```
